```python
# 4.1
secret = int(input("Please enter your secret number: "))
guess = int(input("Please enter your guess of the secret number: "))

if guess > secret:
    print(f"{guess} is to high")
elif guess == secret:
    print(f"{guess} is just right")
else:
    print(f"{guess} is to low")
```

```python
# 4.2
# This program will help you identfy fruits and vegetables
small = input("Is your item small? ")
if small == "yes":
    small = True
else:
    small = False
green = input("Is your item green? ")
if green == "yes":
    green = True
else:
    green = False

print(f"{small} {green}")

if small and green == True:
    print("Your item is a pea")
elif small and green == False:
    print("Your item is a pumpkin")
elif small is True and green is False:
    print("Your item is a cherry")
elif small is False and green is True:
    print("Your item is watermelon")
```
